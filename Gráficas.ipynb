Código para las componentes del tensor de cizalladura $σ_{\mu\nu}$

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def plot_shear_evolution(alpha_deg, Sigma, p, t_max):
    # Rango temporal
    t = np.linspace(0.1, t_max, 500)

    # Factor de escala a(t)
    a = t**p

    # Ángulo en radianes
    alpha = np.radians(alpha_deg)

    # Constantes de anisotropía Σ_i
    Sigma_1 = (2 / np.sqrt(3)) * Sigma * np.sin(alpha)
    Sigma_2 = (2 / np.sqrt(3)) * Sigma * np.sin(alpha + 2 * np.pi / 3)
    Sigma_3 = (2 / np.sqrt(3)) * Sigma * np.sin(alpha + 4 * np.pi / 3)

    # Evolución temporal de las componentes del shear σ^i_j(t) = Σ_i / a(t)^3
    sigma_1 = Sigma_1 / (a**3)
    sigma_2 = Sigma_2 / (a**3)
    sigma_3 = Sigma_3 / (a**3)

    # Escalar de shear σ² = (1/2) σ^i_j σ^j_i = (1/2) (σ₁² + σ₂² + σ₃²)
    # sigma_squared = 0.5 * (sigma_1**2 + sigma_2**2 + sigma_3**2)

    # Figura con una sola gráfica
    fig, ax = plt.subplots(figsize=(7, 4.5))

    # Gráfica 1: Componentes σ^i_j(t)
    ax.plot(t, sigma_1, label=r'$\sigma^{1}{}_{1}(t)$', color='red', linewidth=2)
    ax.plot(t, sigma_2, label=r'$\sigma^{2}{}_{2}(t)$', color='blue', linewidth=2)
    ax.plot(t, sigma_3, label=r'$\sigma^{3}{}_{3}(t)$', color='green', linewidth=2)
    ax.axhline(0, color='black', linestyle='--', alpha=0.6, linewidth=0.8)

    ax.set_xlabel('Tiempo $(t)$', fontsize=11)
    ax.set_ylabel(r'Componentes $\sigma^{i}{}_{j}(t)$', fontsize=11)
    #ax.set_title(rf'Evolución de las componentes ($\sigma^i_j \propto t^{{-3p}}$)', fontsize=11)
    ax.set_title(
        rf"$\alpha = {alpha_deg:.1f}^\circ$, $\Sigma = {Sigma:.2f}$",
        fontsize=11,
        pad=12,
    )
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='upper right', frameon=True)

    # --- Gráfica 2: Escalar de Shear σ²(t) (COMENTADA) ---
    # ax2.plot(t, sigma_squared, label=r'$\sigma^2(t)$', color='#d62728', linewidth=2)
    # ax2.set_xlabel('Tiempo $t$', fontsize=11)
    # ax2.set_ylabel(r'Escalar de Shear $\sigma^2$', fontsize=11)
    # ax2.set_title(r'Decaimiento del Escalar de Anisotropía $\sigma^2(t)$', fontsize=11)
    # ax2.grid(True, linestyle=':', alpha=0.6)
    # ax2.legend(loc='upper right', frameon=True)

    plt.tight_layout()
    plt.savefig('evolucion_shear.png', dpi=300, bbox_inches='tight')

    plt.show()

# Controles interactivos
alpha_slider = widgets.FloatSlider(
    value=30.0, min=0.0, max=360.0, step=1.0,
    description=r'$\alpha$ (deg):', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px')
)

Sigma_slider = widgets.FloatSlider(
    value=1.0, min=0.0, max=5.0, step=0.1,
    description=r'$\Sigma$:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px')
)

p_slider = widgets.FloatSlider(
    value=0.667, min=0.1, max=2.0, step=0.05,
    description=r'Potencia $p$ ($a(t)=t^p$):', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px')
)

t_max_slider = widgets.FloatSlider(
    value=5.0, min=1.0, max=20.0, step=0.5,
    description=r'$t_{{\max}}$:', style={'description_width': 'initial'}, layout=widgets.Layout(width='400px')
)

ui = widgets.VBox([alpha_slider, Sigma_slider, p_slider, t_max_slider])
out = widgets.interactive_output(
    plot_shear_evolution,
    {'alpha_deg': alpha_slider, 'Sigma': Sigma_slider, 'p': p_slider, 't_max': t_max_slider}
)

display(ui, out)
```

Código para los factores de escala $a_{i}(t)$ y el factor de escala medio $a(t)$

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def graficar_nueva_solucion(alpha=0.2, M=1.0, Sigma=1.0):


    # Convertir ángulo de grados a radianes
    alpha
    #  Dominio temporal
    t = np.linspace(0.01, 5, 2000)

    # Factor de escala promedio a(t)
    a3 = (3/4) * M * (t**2) + np.sqrt(3) * Sigma * t
    a = a3**(1/3)

    W = (1 / np.sqrt(3)*Sigma) * np.log( t / (t + (4*Sigma)/(np.sqrt(3)*M) ))

    #  Constantes de cizalladura direccionales Sigma_i(alpha)
    Sigma1 = (2/np.sqrt(3)) * Sigma * np.sin(alpha)
    Sigma2 = (2/np.sqrt(3)) * Sigma * np.sin(alpha + 2*np.pi/3)
    Sigma3 = (2/np.sqrt(3)) * Sigma * np.sin(alpha + 4*np.pi/3)

    #  Factores de escala direccionales a_i(t)
    a1 = a * np.exp(Sigma1 * W)
    a2 = a * np.exp(Sigma2 * W)
    a3_dir = a * np.exp(Sigma3 * W)

    # 6. Gráfica
    plt.figure(figsize=(10, 6))

    plt.plot(t, a, 'k--', linewidth=2.5, label=r'$a(t)$ (Medio)')
    plt.plot(t, a1, 'r-', linewidth=2, label=r'$a_1(t)$')
    plt.plot(t, a2, 'g-', linewidth=2, label=r'$a_2(t)$')
    plt.plot(t, a3_dir, 'b-', linewidth=2, label=r'$a_3(t)$')

    plt.title(#f'Evolución de Factores de Escala (Nueva $W(t)$)\n'
              f'$\\alpha = {alpha:.2f}$, $M = {M:.2f}$, $\\Sigma = {Sigma:.2f}$', fontsize=13)
    plt.xlabel('Tiempo $(t)$', fontsize=12)
    plt.ylabel('Factor de escala $a_{i}(t)$', fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(fontsize=11, loc='upper left')
    plt.tight_layout()
    plt.show()

# Controles interactivos
interact(
    graficar_nueva_solucion,
    alpha=FloatSlider(min=-np.pi, max=np.pi, step=0.01, value=0.2, description=r'$\alpha$'),
    M=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description=r'$M$'),
    Sigma=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description=r'$\Sigma$')
);
```

Tasas de expansión direccionales $H_{i}(t)$

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def graficar_tasas_tu_solucion(alpha_deg=30.0, M=1.0, Sigma=1.0):

    # 1. Convertir ángulo de grados a radianes
    alpha = np.radians(alpha_deg)

    # 2. Dominio temporal (t > 0)
    t = np.linspace(0.01, 0.4, 2000)

    # 3. Ángulos direccionales alpha_i
    alpha1 = alpha
    alpha2 = alpha + (2 * np.pi / 3)
    alpha3 = alpha + (4 * np.pi / 3)

    # 4. Cálculo de H_i(t) con tu fórmula exacta
    denominador = (M / 4) * t + (Sigma / np.sqrt(3))

    num1 = (M / 4) * t + (Sigma / (2 * np.sqrt(3))) * (1 + 2 * np.sin(alpha1))
    num2 = (M / 4) * t + (Sigma / (2 * np.sqrt(3))) * (1 + 2 * np.sin(alpha2))
    num3 = (M / 4) * t + (Sigma / (2 * np.sqrt(3))) * (1 + 2 * np.sin(alpha3))

    H1 = (2 / (3 * t)) * (num1 / denominador)
    H2 = (2 / (3 * t)) * (num2 / denominador)
    H3 = (2 / (3 * t)) * (num3 / denominador)

    # 5. Tasa de expansión media H(t) = (1/3) * (H1 + H2 + H3)
    H_medio = (H1 + H2 + H3) / 3.0

    # 6. Gráfica
    plt.figure(figsize=(10, 6))

    plt.plot(t, H_medio, 'k--', linewidth=2.5, label=r'$H(t)$ (Medio)')
    plt.plot(t, H1, 'r-', linewidth=2, label=r'$H_1(t)$')
    plt.plot(t, H2, 'g-', linewidth=2, label=r'$H_2(t)$')
    plt.plot(t, H3, 'b-', linewidth=2, label=r'$H_3(t)$')

    plt.axhline(0, color='gray', linestyle=':', alpha=0.7)

    plt.title(#f'Tasas de Expansión Direccionales (Tu Solución Exacta)\n'
              f'$\\alpha = {alpha_deg:.1f}^\\circ$, $M = {M:.2f}$, $\\Sigma = {Sigma:.2f}$', fontsize=13)
    plt.xlabel('Tiempo ($t$)', fontsize=12)
    plt.ylabel('Tasa de expansión $H_i(t)$', fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend(fontsize=11, loc='upper right')

    # Guardar en alta calidad (PNG y PDF)
    plt.savefig('tasas_expansion_tu_solucion.png', dpi=300, bbox_inches='tight')
    plt.savefig('tasas_expansion_tu_solucion.pdf', bbox_inches='tight')

    plt.tight_layout()
    plt.show()

# Controles interactivos
interact(
    graficar_tasas_tu_solucion,
    alpha_deg=FloatSlider(min=-30, max=90, step=1.0, value=30.0, description=r'$\alpha (^\circ)$'),
    M=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description=r'$M$'),
    Sigma=FloatSlider(min=0.1, max=5.0, step=0.1, value=1.0, description=r'$\Sigma$')
);
```

Relación de las tasas de expansión direccionales $\frac{H_{i}(t)}{H(t)}$

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def ratio_Hi_H(t, M, Sigma, alpha_i):
    """Calcula la relación de expansión direccional Hi(t)/H(t)."""
    num_Hi = (M / 4.0) * t + (Sigma / (2.0 * np.sqrt(3))) * (1.0 + 2.0 * np.sin(alpha_i))
    den_Hi = (M / 4.0) * t + (Sigma / np.sqrt(3))
    Hi = num_Hi / den_Hi

    num_H = (3.0 * M / 4.0) * t + (np.sqrt(3) * Sigma / 2.0)
    den_H = (3.0 * M / 4.0) * t + (np.sqrt(3) * Sigma)
    H = num_H / den_H

    return Hi / H

def plot_directional_evolution(M, Sigma, alpha):
    t = np.linspace(0.01, 10.0, 500)

    # Angulos para las tres direcciones espacialmente anisotrópicas
    a1 = alpha
    a2 = alpha + 2.0 * np.pi / 3.0
    a3 = alpha + 4.0 * np.pi / 3.0

    # Tasas de expansión direccionales normalizadas
    r1 = ratio_Hi_H(t, M, Sigma, a1)
    r2 = ratio_Hi_H(t, M, Sigma, a2)
    r3 = ratio_Hi_H(t, M, Sigma, a3)

    plt.figure(figsize=(9, 5))
    plt.plot(t, r1, color='red', lw=2, label=r'$H_1/H $')
    plt.plot(t, r2, color='green', lw=2, label=r'$H_2/H $')
    plt.plot(t, r3, color='blue', lw=2, label=r'$H_3/H $')

    # Línea de isotropía
    plt.axhline(1.0, color='gray', linestyle='--', alpha=0.7, label='$H_i/H = 1$')

    plt.xlim(0, 10)
    plt.xlabel(r'Tiempo $(t)$', fontsize=11)
    plt.ylabel(r'$H_i(t) / H(t)$', fontsize=11)
    plt.title(f'$\\alpha={alpha:.3f}, \\ M={M}, \\Sigma={Sigma}$', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend(loc='upper right')
    plt.savefig('expansion_evolucion', dpi=300, bbox_inches='tight')

    plt.tight_layout()
    plt.show()

# Deslizadores para M, Sigma y el ángulo alpha
interact(
    plot_directional_evolution,
    M=widgets.FloatSlider(value=1.0, min=0.0, max=5.0, step=0.1, description='M'),
    Sigma=widgets.FloatSlider(value=1.0, min=0.0, max=5.0, step=0.1, description='Σ'),
    alpha=widgets.FloatSlider(
        value=0.0,
        min=-np.pi/6,
        max=np.pi/2,
        step=0.01,
        description='α (rad)'
    )
);
```
