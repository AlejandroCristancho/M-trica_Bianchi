```python
%display latex

M = Manifold(4, 'M', r'\mathcal{M}')
X.<t, x, y, z> = M.chart(r't x y z')

var('G', domain='real')
var('Lambda', domain='real')
var('w', domain='real')  # ecuación de estado p = w ρ

a1 = function('a1')(t)
a2 = function('a2')(t)
a3 = function('a3')(t)
rho = function('rho')(t)
p   = function('p')(t)

g = M.metric('g', signature=-2)
g[0,0] =  1
g[1,1] = -a1^2
g[2,2] = -a2^2
g[3,3] = -a3^2

# 4-velocidad comóvil
u = M.vector_field(name='u')
u[0] = 1
u_form = u.down(g)

# Tensor energía-momento
T = (rho + p)*(u_form*u_form) - p*g
show(T.display())
```

```python
# Conexión
nab = g.connection()

# ∇_μ T^{μν}  —  calcular componente ν=0
# Primero, subir índice con la métrica inversa
T_up = T.up(g)   # T^{μν}

# Divergencia
divT = nab(T_up).trace(0, 2)   # contrae el primer y tercer índice

print("∇_μ T^{μν}:")
divT.display()

# Componente ν=0 → ecuación de continuidad
print("\nComponente ν=0:")
show(divT[0].expr().simplify_full())
```

```python
# Definir V = a1*a2*a3
V = a1*a2*a3

# La solución ρ = ρ0 V^{-(1+w)}
rho0 = var('rho0', domain='positive')
rho_prop = rho0 * V^(-(1+w))

# Verificar que satisface la ecuación de continuidad
continuidad = diff(rho_prop, t) + (diff(a1,t)/a1 + diff(a2,t)/a2 + diff(a3,t)/a3)*(rho_prop*(1+w))
show(continuidad.simplify_full())
```
